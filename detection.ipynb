```python
from ultralytics import YOLO
import os

# 1. Load your trained model
# If you just finished training, it's in 'runs/detect/train/weights/best.pt'
# If you've restarted the session, make sure you've uploaded your best.pt file
model_path = 'runs/detect/train/weights/best.pt'

if os.path.exists(model_path):
    model = YOLO(model_path)
    
    # 2. Define the path to your dataset video
    video_input_path = '/kaggle/input/roaaddd/detected_road_hazards.mp4'

    # 3. Run Inference
    # we use 'save=True' to generate the output video with boxes
    # we use 'conf=0.3' to show only detections with 30% confidence or higher
    results = model.predict(source=video_input_path, save=True, conf=0.3)

    print(f"Success! The processed video is saved in: {results[0].save_dir}")
else:
    print("Model file not found. Please check the path to your 'best.pt' file.")
```

```python
import matplotlib.pyplot as plt
from collections import Counter

# 1. Collect all detections from the video inference results
# 'results' is the variable from your previous model.predict() call
all_labels = []
for r in results:
    for c in r.boxes.cls:
        label = model.names[int(c)]
        all_labels.append(label)

# 2. Count occurrences
counts = Counter(all_labels)

# 3. Create the Plot
plt.figure(figsize=(10, 6))
plt.bar(counts.keys(), counts.values(), color='skyblue', edgecolor='navy')
plt.xlabel('Hazard Type', fontsize=12)
plt.ylabel('Total Detections (Frames)', fontsize=12)
plt.title('Object Detection Summary for Road Hazards Video', fontsize=14)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
